# Carregar, netejar i filtrar DADES 1

```python
#Definir periode de mostreix en milisegons:
mostra = mostra/1000
# =========================
# BLOC 0: definicions
# =========================
DADES_2_EN = 0

def filtre_passa_baixos(serie, fs=(1/mostra), fc=0.1, order=3):
    """
    Filtra una sèrie temporal amb un filtre Butterworth passa baixos.
    fs: freqüència de mostreig (Hz) — 1/0.1s = 10 Hz
    fc: freqüència de tall (Hz)
    order: ordre del filtre
    """
    nyq = fs / 2
    normal_cutoff = fc / nyq
    b, a = signal.butter(order, normal_cutoff, btype='low', analog=False)
    return signal.filtfilt(b, a, serie)

# =========================
# BLOC 1: Carregar Dades 1, netejar i filtrar
# =========================

#file1 = "Niche1.csv"  # canviar-ho si cal
label1 = os.path.splitext(os.path.basename(file1))[0]
full_path = os.path.join(base_path, file1)
dades1 = pd.read_csv(full_path, sep=';', on_bad_lines='skip', decimal='.')  #pensar en posar un .
dades1["temps"] = np.arange(len(dades1)) * mostra + 0
if "NTC0" in dades1:
  dades1["NTC0_filt"] = filtre_passa_baixos(dades1["NTC0"])
if "NTC1" in dades1:
  dades1["NTC1_filt"] = filtre_passa_baixos(dades1["NTC1"])
if "NTC2" in dades1:
  dades1["NTC2_filt"] = filtre_passa_baixos(dades1["NTC2"])
if "NTCA" in dades1:
  dades1["NTCA_filt"] = filtre_passa_baixos(dades1["NTCA"])
if "PWR_GEN" in dades1:
  dades1["PWR_GEN"] = -dades1["PWR_GEN"]
if "RPM_GEN" in dades1:
  dades1["CAD_GEN"] = dades1["RPM_GEN"]/56.06
dades1.head()




```

```python
# Configuració
tot_dataset = True
#TODO que es digui segon o temps incial
mostra_inicial = 2228
mostra_final = 2352
#TODO arreglar les equacions epr quan no sigui 500ms
mostra_inicial = mostra_inicial*2
mostra_final = mostra_final*2
df = dades1[["temps", "PWR_BAT"]].copy()

# Seleccionar el rang de mostres abans de netejar les dades
if not tot_dataset:
    if not (1 <= mostra_inicial < mostra_final <= len(df)):
        raise ValueError(
            f"Cal complir: 1 <= mostra_inicial < mostra_final <= {len(df)}"
        )

    # iloc comença a 0 i exclou el límit final:
    # així incloem les mostres 100 a 1000.
    df = df.iloc[mostra_inicial - 1:mostra_final].copy()
print(df)
df["temps"] = pd.to_numeric(df["temps"], errors="coerce")
df["PWR_BAT"] = pd.to_numeric(df["PWR_BAT"], errors="coerce")
df = df.dropna().sort_values("temps")

if len(df) < 2:
    raise ValueError("Calen almenys dues mostres vàlides per calcular l'energia.")

t = df["temps"].to_numpy()
p = df["PWR_BAT"].to_numpy()

# Separar consum i regeneració
p_consum = np.clip(p, 0, None)      # només valors positius
p_regen = np.clip(-p, 0, None)      # negatius passats a positius
print(p)
# Integració -> Wh
energia_gastada_Wh = np.trapezoid(p_consum, t) / 3600
energia_recuperada_Wh = np.trapezoid(p_regen, t) / 3600
energia_neta_Wh = energia_gastada_Wh - energia_recuperada_Wh

if tot_dataset:
    print("Selecció: tot el dataset")
else:
    print(f"Selecció: mostra {mostra_inicial} a mostra {mostra_final}")

print(f"Energia gastada: {energia_gastada_Wh:.3f} Wh")
print(f"Energia recuperada: {energia_recuperada_Wh:.3f} Wh")
print(f"Energia neta: {energia_neta_Wh:.3f} Wh")
print(f"percentatge recuperat: {energia_recuperada_Wh/energia_gastada_Wh*100 :.1f} %")
```

Calcul de mitjanes

```python
# ─── MITJANES PER INTERVAL ─────────────────────────────────────────────────────
t_inici = 3100    # ← modifica aquí
t_fi    = 3190  # ← modifica aquí
# ───────────────────────────────────────────────────────────────────────────────

def mitjanes_interval(df, label, t_inici, t_fi):
    mascara = (df["temps"] >= t_inici) & (df["temps"] <= t_fi)
    df_tall = df.loc[mascara]
    n_files = len(df_tall)
    if n_files == 0:
        print(f"⚠️  [{label}] Cap fila en l'interval [{t_inici}, {t_fi}]")
        return
    cols_num = [c for c in df_tall.select_dtypes(include='number').columns if c != "temps"]
    resultat = pd.DataFrame(
        {"Mitjana": df_tall[cols_num].mean(),
         "Màxim":   df_tall[cols_num].max(),
         "Mínim":   df_tall[cols_num].min()},
        index=cols_num
    ).T
    print(f"\n📊 {label}  ({n_files} mostres) | interval [{t_inici}s, {t_fi}s] temps: {(t_fi-t_inici)/60}min")
    display(resultat.style.format("{:.4f}"))

mitjanes_interval(dades1, label1, t_inici, t_fi)
if DADES_2_EN:
    mitjanes_interval(dades2, label2, t_inici, t_fi)
```
